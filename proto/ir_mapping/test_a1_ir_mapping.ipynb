```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import measure

# Example: Create a sample function f(x, y) = x^2 + y^2 - 1
x = np.linspace(-1.5, 1.5, 100)
y = np.linspace(-1.5, 1.5, 100)
xx, yy = np.meshgrid(x, y)
M = xx**2 + yy**2 - 1  # Implicit function f(x, y) = 0 defines a unit circle

# Extract the contour for f(x, y) = 0
contours = measure.find_contours(M, level=0)
for contour in contours:
    contour[:, 0] = contour[:, 0] * (x.max() - x.min()) / (len(x) - 1) + x.min()
    contour[:, 1] = contour[:, 1] * (y.max() - y.min()) / (len(y) - 1) + y.min()

# Plot the extracted contour
plt.figure(figsize=(6, 6))
ext = (x.min(), x.max(), y.min(), y.max())
plt.contourf(xx, yy, M, levels=20, cmap="coolwarm", alpha=0.5, extent=ext)  # Background field
for contour in contours:
    plt.plot(contour[:, 1], contour[:, 0], 'k', linewidth=2)  # Contours are in (row, col) format

plt.xlabel("x")
plt.ylabel("y")
plt.title("Contour line for f(x, y) = 0")
plt.axis("equal")
plt.show()
```
